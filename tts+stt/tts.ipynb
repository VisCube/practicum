```python
import edge_tts

TEXT = """
Просто набор букв
"""

OUTPUT_FILE = "russian_voice.mp3"

voice = "ru-RU-SvetlanaNeural"

communicate = edge_tts.Communicate(
    TEXT,
    voice,
    rate="+0%",
    volume="+0%"
)

await communicate.save(OUTPUT_FILE)

print("Готово!")
print(f"Файл сохранён: {OUTPUT_FILE}")
```

```python
import torch
import soundfile as sf

result = torch.hub.load(
    repo_or_dir="snakers4/silero-models",
    model="silero_tts",
    language="ru",
    speaker="v5_5_ru",
    trust_repo=True
)

model = result[0]
print("Модель загружена.")

text = """
Какая-то белиберда
"""

audio = model.apply_tts(
    text=text,
    speaker="eugene",
    sample_rate=48000
)

sf.write(
    "russian_voice.wav",
    audio.numpy(),
    48000
)

print("Готово!")
print("Файл: russian_voice.wav")
```

```python
import wave
from piper import PiperVoice

TEXT = """
Здесь мог быть полноценный ответ
"""

MODEL = "ru_RU-irina-medium.onnx"
OUTPUT = "russian_voice.wav"

voice = PiperVoice.load(MODEL)

print("Генерация...")

with wave.open(OUTPUT, "wb") as wav_file:
    voice.synthesize_wav(TEXT, wav_file)

print("Готово!")
print(f"Файл: {OUTPUT}")
```

```python
import os
import urllib.request
import soundfile as sf
from kokoro_onnx import Kokoro


MODEL_URL = (
    "https://github.com/thewh1teagle/kokoro-onnx/"
    "releases/download/model-files-v1.1/kokoro-v1.0.onnx"
)

VOICES_URL = (
    "https://github.com/thewh1teagle/kokoro-onnx/"
    "releases/download/model-files-v1.1/voices-v1.0.bin"
)

MODEL_FILE = "kokoro-v1.0.onnx"
VOICES_FILE = "voices-v1.0.bin"


def download_file(url, filename):
    if os.path.exists(filename):
        print(f"Файл уже существует: {filename}")
        return

    print(f"Скачивание {filename}...")

    urllib.request.urlretrieve(url, filename)

    print(f"Готово: {filename}")



download_file(MODEL_URL, MODEL_FILE)
download_file(VOICES_URL, VOICES_FILE)


kokoro = Kokoro(
    MODEL_FILE,
    VOICES_FILE
)

text_ru = (
    "Наверное, что-то осмысленное"
)

print("Генерация речи...")

audio, sample_rate = kokoro.create(
    text_ru,
    voice="af_sarah",
    speed=1.0
)

output_file = "russian_voice.wav"

sf.write(
    output_file,
    audio,
    sample_rate
)

print()
print("Готово!")
print(f"Файл сохранён: {output_file}")
```
